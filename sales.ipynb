```python
import sqlite3
import pandas as pd
filename = "SQL_Sales_Dataset_200_Rows.xlsx"
df = pd.read_excel(filename)

conn = sqlite3.connect(':memory:')

df.to_sql('sales', conn, index=False, if_exists='replace')
```

```python
#regions with highest revenue

query = """
    SELECT region, SUM(total_price) AS total_revenue
    FROM sales
    GROUP BY region
    ORDER BY total_revenue DESC;
"""
display(pd.read_sql(query, conn))
```

```python
#highest price orders
query = """
    SELECT order_id, customer_name, product_name, total_price
    FROM sales
    ORDER BY total_price DESC
    LIMIT 5;
"""
display(pd.read_sql(query, conn))
```

```python
#Average price and total orders according to category
query = """
    SELECT category,
           COUNT(order_id) AS total_orders,
           AVG(unit_price) AS average_unit_price
    FROM sales
    GROUP BY category;
"""
display(pd.read_sql(query, conn))
```

```python
#East regions with category either Electronics or Furniture
query = """
    SELECT order_id, customer_name, category, sub_category, total_price, region
    FROM sales
    WHERE region = 'East'
      AND (category = 'Electronics' OR category = 'Furniture')
    ORDER BY total_price DESC;
"""
display(pd.read_sql(query, conn))
```

```python
#total quantity, min price and max price
query = """
    SELECT sub_category,
           SUM(quantity) AS total_quantity,
           MIN(unit_price) AS min_unit_price,
           MAX(unit_price) AS max_unit_price
    FROM sales
    GROUP BY sub_category
    ORDER BY total_quantity DESC;
"""
display(pd.read_sql(query, conn))
```

```python
query = """
    SELECT category, SUM(total_price) AS total_revenue
    FROM sales
    GROUP BY category
    HAVING SUM(total_price) > 100000
    ORDER BY total_revenue DESC;
"""
display(pd.read_sql(query, conn))
```

```python
query = """
    SELECT DISTINCT customer_name, category, region
    FROM sales
    WHERE customer_name LIKE 'M%'
    ORDER BY customer_name;
"""
display(pd.read_sql(query, conn))
```

```python
query = """
    SELECT order_id, customer_name, order_date, total_price
    FROM sales
    WHERE order_date LIKE '2025-05%'
    ORDER BY order_date ASC;
"""
display(pd.read_sql(query, conn))
```

```python
query = """
    SELECT sub_category,
           SUM(quantity) AS total_quantity,
           MIN(unit_price) AS min_unit_price,
           MAX(unit_price) AS max_unit_price
    FROM sales
    GROUP BY sub_category
    ORDER BY total_quantity DESC;
"""
display(pd.read_sql(query, conn))
```

```python
# Create an 'orders' table and a separate 'customers' table to practice joins
query_orders = """
    CREATE TABLE orders_subset AS
    SELECT order_id, customer_name, order_date, total_price, region
    FROM sales;
"""

query_customers = """
    CREATE TABLE customer_info AS
    SELECT DISTINCT customer_name, region
    FROM sales;
"""

conn.execute("DROP TABLE IF EXISTS orders_subset;")
conn.execute("DROP TABLE IF EXISTS customer_info;")
conn.execute(query_orders)
conn.execute(query_customers)

print("Tables 'orders_subset' and 'customer_info' created successfully!")
```

```python
query = """
    SELECT o.order_id, c.customer_name, o.order_date, o.total_price, o.region
    FROM orders_subset o
    INNER JOIN customer_info c ON o.customer_name = c.customer_name
    LIMIT 10;
"""
display(pd.read_sql(query, conn))
```

```python
query = """
    SELECT c.customer_name, o.order_id, o.total_price
    FROM customer_info c
    LEFT JOIN orders_subset o ON c.customer_name = o.customer_name
    LIMIT 10;
"""
display(pd.read_sql(query, conn))
```

```python
query = """
    SELECT order_id, customer_name, product_name, total_price
    FROM sales
    WHERE total_price > (
        SELECT AVG(total_price)
        FROM sales
    )
    ORDER BY total_price DESC;
"""
display(pd.read_sql(query, conn))
```

```python
query = """
    SELECT order_id, customer_name, category, product_name, total_price
    FROM sales
    WHERE customer_name IN (
        SELECT DISTINCT customer_name
        FROM sales
        WHERE category = 'Electronics'
    )
    ORDER BY customer_name;
"""
display(pd.read_sql(query, conn))
```
